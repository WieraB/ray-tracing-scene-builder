```python

from netgen.occ import *
from ngsolve import Mesh
from ngsolve.webgui import Draw

#  = Sphere((0,0,0), 2)
s = Sphere((-0.006, -5.0, -1.), 6)
geo = OCCGeometry(s)
mesh = geo.GenerateMesh()
mesh.SecondOrder()
# mesh.Export("sphere.msh", "Gmsh Format")
mesh.Save("sphere.vol")
mesh = Mesh(mesh)
Draw(mesh, subdivision=10)

# print(mesh.GetCurveOrder())
```

```python
from netgen.csg import *

cube = OrthoBrick( Pnt(0,0,0), Pnt(1,1,1) )
hole = Cylinder ( Pnt(0.5, 0.5, 0), Pnt(0.5, 0.5, 1), 0.2)

s1 = Sphere((-0.006, -5.0, -1.), 6)

s2 = Sphere((4, -5.0, -1.), 3.5)

geo = CSGeometry()
geo.Add (s1-s2)

mesh = geo.GenerateMesh()
mesh.SecondOrder()

mesh.Save("sphere2.vol")

mesh = Mesh(mesh)
Draw(mesh, subdivision=10)
```
