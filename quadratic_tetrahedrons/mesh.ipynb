```python

from netgen.occ import *
from ngsolve import Mesh
from ngsolve.webgui import Draw

s = Sphere((0,0,0), 2)
geo = OCCGeometry(s)
mesh = geo.GenerateMesh()
mesh.SecondOrder()
# mesh.Export("sphere.msh", "Gmsh Format")
mesh.Save("sphere.vol")
mesh = Mesh(mesh)
Draw(mesh, subdivision=10)

print(mesh.GetCurveOrder())
```
